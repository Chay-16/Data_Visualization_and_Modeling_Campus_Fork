## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
characters_df.groupby("Universe")["Appearances"].mean()
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: using .groupby with the "Universe" column then using the column "appearances" in brackets with a .mean() gave the average number of appearances per universe and the data shows DC has a higher average of per character appearances with 23.7875159 compared to Marvel's average of 17.360109


### Step 4: Check

```python
# Optional: add a targeted sanity check if code would help.
characters_df.groupby("Universe")["Appearances"].agg(['count', 'min', 'median', 'max', 'std'])
```

**What would make you suspicious, and why? Does your evidence make sense in context? Describe any issue you found and how you addressed it.**

Your answer: The evidence seems to make sense in context of comic book appearances as the min for both is 1 meaning there isn't a nonsensical negative value that could affect the data, the median for DC is 6.0 and the median for Marvel is 3.0, the overall count of characters for DC is lower at 6479 compared to Marvels 14665 however the counts are large enough by themselves that the difference isn't enough to greatly impact the data. One thing I did find interesting about the data was that Marvel actually has a higher max number of appearances at 4043 compared to DC's max of 3093 but DC has a lower standard deviation of 87.774133 compared to Marvels 97.183428 which, when combined with the median, helps explain how DC was able to reach a higher average. The max number of appearances being so low did also show that the appearances count come from a smaller time frame but the numbers from both being similar also means that the same (or at least a very similar) observed time frame was used for both universes

### Step 5: Conclusion


**Answer your research question in a sentence using the numerical evidence. How does it compare with your prediction?**

Your answer: DC has the highest average number of appearances with 23.787159 compared to Marvels 17.360109. This matches my predicition made before calculation.

**What should a reader avoid concluding from this result, and why? Connect your limitation to this question and dataset.**

## Investigation 3

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 3?**

Your answer: Are the majority of characters with less than 15 appearances still alive?


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer: I would predict no as I think many of those characters could be introduced just to have their deaths used to add drama to the main heroes story.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer: For this question, I'll filter the data set to only be characters with less than 15 appearances, then calculate the percentage of characters within the filtered dataset that are alive using value_counts with normalize to divide the number of each category by the total number of characters. There were no missing values for this dataset, any non-relevant characters will be filtered out, and there are only 2 options of either true of false for alive status.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# Prepare, filter, or calculate features as needed; display useful intermediate results.
filtered_characters = characters_df[characters_df["Appearances"].between (0, 15)] 
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
filtered_characters["Alive"].value_counts(normalize=True)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: The Alive percentage being about 76% (.763374) for yes/true and about 24% (0.236626) for no/false anwsers the question of whether a majority of characters with less than 15 appearnces are still alive

With how many characters were logged at less than 15 appearances, the evidence and percentage breakdown makes sense in context.

**Answer your research question in a sentence using the numerical evidence. How does it compare with your prediction?**
